## 1. Research Question

- **Main research question:** Is daily social media use associated with self-reported stress levels among young adults?
- **Why is this question interesting or important?** Social media is a huge part of daily life for most young adults, and stress is linked to sleep, wellbeing and performance at study or work. A 2025 systematic review of 57 studies (571,427 participants) found that daily social media use is associated with increased stress, anxiety, depression and poor sleep quality, but individual studies disagree and most evidence is correlational, so the question is worth testing directly on a fresh dataset.
- **Who might use or care about the answer?** Young adults, university and workplaces, wellbeing services, parents and educators, mental health practitioners, platform designers, policymakers working on digital wellbeing, and other researchers.
- **Possible sub-questions:**
  1. How are `Avg_Daily_Usage_Hours` and `Stress_Level` distributed across the 5,000 students (range, skew, outliers)?
  2. Is the association between usage hours and stress consistent across academic level (High School/Undergraduate/Graduate), gender, and country?
  3. Does `Daily_Unlocks` (checking frequency) track stress independently of total hours used, or does it just duplicate the same signal?
  4. Does the relationship persist after accounting for `Sleep_Hours_Per_Night` and `Study_Hours`, which both also correlate strongly with stress?
  5. Does `Purpose_Of_Use` (Entertainment vs Education vs Networking vs News) change the strength of the association?
